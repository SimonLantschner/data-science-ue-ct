```python
import pandas as pd
from matplotlib import pyplot as plt
from pathlib import Path
```

```python
# Damit wir Daten laden können, müssen wir zueerst in Colab hochladen


# Pfad zur Datei (relativ zum Notebook)
csv_path = Path("../data/titanic.csv")

df = pd.read_csv(
    csv_path,
    sep=",",
    low_memory=False, 
    encoding="utf-8",
)

print(df.shape)
print(df.dtypes)
df.head()
```

```python
df.info() # Gibt eine Zusammenfassung der DataFrame-Informationen aus, einschließlich der Anzahl der Einträge, Spaltennamen, Datentypen und Speicherverbrauch.
```

```python
df.describe() # Gibt eine statistische Zusammenfassung der numerischen Spalten im DataFrame zurück, einschließlich Anzahl, Mittelwert, Standardabweichung, Minimum, 25. Perzentil, Median (50. Perzentil), 75. Perzentil und Maximum.
```

```python
df["Pclass"].plot(kind="hist") # Erstellt ein Histogramm für die angegebene Spalte im DataFrame. Die x-Achse zeigt die Werte der Spalte, und die y-Achse zeigt die Häufigkeit dieser Werte.
```

```python
df["Pclass"].value_counts().plot(kind="barh") # Erstellt ein Balkendiagramm für die Häufigkeit der Werte in der angegebenen Spalte. Die x-Achse zeigt die eindeutigen Werte der Spalte, und die y-Achse zeigt die Anzahl der Vorkommen jedes Wertes.
```

```python
df["Age"].value_counts().sort_index().plot(kind="barh", figsize=(6, 20)) #Wie groß soll das figure sein (Höhe und Breite)
plt.title("Distribution of Age")
plt.xlabel("Number of persons")
plt.ylabel("Age Group")
plt.savefig("../image/age_plot.png")
```
