```python
from pandas import read_csv
from sklearn.model_selection import train_test_split
```

```python
data = read_csv("https://www.tinyurl.com/mcit-titanic")
```

## Daten Vertical splitten (Input/Output)

```python
X = data[[
    'Pclass',
    'Sex',
    'Age',
    'SibSp',
    'Parch',
    'Embarked'
 ]]

y = data[[
    'Survived'
 ]]
```

## Daten horizontal Splitten (Train/Test)

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
print("Trainings-Daten", X_train.shape, y_train.shape)
print("Test-Daten", X_test.shape, y_test.shape)
```
